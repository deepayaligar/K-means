## K -means cluster

#### identifies K cluster(grouping) within the dataset

#### steps :


> Choose 3(based on k value)  random points as  Centroids(mean)

> Calculate distance bw points and centroids and than assign points to cloest centroid ..

>  compute new centroids ..

note :  we need to repeat step 2 and step 3 until we reach stability (centroids value gets repeated)

```python

from sklearn.model_selection import train_test_split
train, test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.shape
```

```python
test.shape
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(train)
```

```python
predict = model.predict(test)
```

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
sns.scatterplot(x=train['petal length (cm)'],y=train['petal width (cm)'], hue=model.labels_)
```

```python
sns.scatterplot(x=test['petal length (cm)'],y=test['petal width (cm)'],hue=predict)
```
